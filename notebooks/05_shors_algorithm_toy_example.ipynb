# 05 - Shor's Algorithm Toy Example

This notebook explains Shor's algorithm at a conceptual and toy-example level.

It does not implement a practical cryptanalytic attack. The goal is to understand why Shor's algorithm matters for cybersecurity.


## 1. Why Shor Matters

Shor's algorithm can theoretically solve integer factorization and discrete logarithms efficiently on a sufficiently powerful fault-tolerant quantum computer.

This matters because:

- RSA relies on the hardness of integer factorization.
- Diffie-Hellman relies on discrete logarithms.
- Elliptic Curve Cryptography relies on elliptic curve discrete logarithms.


## 2. Toy Factorization Example

A common educational example is factoring 15:

```text
15 = 3 x 5
```

This is trivial classically, but it helps explain the structure of Shor's algorithm.


````python
import math

N = 15
a = 2

g = math.gcd(a, N)
print(f'gcd({a}, {N}) = {g}')
````

If `gcd(a, N)` is greater than 1, a non-trivial factor has already been found. Otherwise, Shor's algorithm uses a quantum period-finding subroutine.


## 3. Classical Period Search for the Toy Example

For learning purposes, we can compute the period classically. The quantum algorithm would use a quantum subroutine to estimate this period efficiently.


````python
def find_period_classically(a: int, N: int, max_r: int = 100) -> int | None:
    for r in range(1, max_r + 1):
        if pow(a, r, N) == 1:
            return r
    return None

r = find_period_classically(a, N)
print(f'Period r = {r}')
````

## 4. Deriving Factors from the Period

If the period `r` is even and the conditions are suitable, factors can be derived from:

```text
gcd(a^(r/2) - 1, N)
gcd(a^(r/2) + 1, N)
```


````python
if r is not None and r % 2 == 0:
    x = pow(a, r // 2, N)
    factor_1 = math.gcd(x - 1, N)
    factor_2 = math.gcd(x + 1, N)
    print(f'Factors derived from period: {factor_1}, {factor_2}')
else:
    print('The period is not suitable for deriving factors in this run.')
````

## 5. Cybersecurity Relevance

Small examples do not break real cryptography. However, Shor's algorithm is strategically important because a large fault-tolerant quantum computer could threaten RSA, Diffie-Hellman, and ECC.

This motivates post-quantum cryptography and long-term migration planning.
