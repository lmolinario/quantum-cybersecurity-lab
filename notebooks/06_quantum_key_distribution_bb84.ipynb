# 06 - Quantum Key Distribution and BB84

This notebook introduces Quantum Key Distribution (QKD) and the BB84 protocol using a small classical simulation.

The goal is educational: understand the protocol flow and why measurement disturbance matters.


## 1. BB84 Actors

Traditional BB84 explanations use three actors:

- Alice: sender
- Bob: receiver
- Eve: potential eavesdropper

Alice sends quantum states. Bob measures them using randomly selected bases. Alice and Bob later compare bases and keep only compatible measurements.


```python
import random

random.seed(42)

NUM_BITS = 32
BASES = ['+', 'x']  # '+' = rectilinear, 'x' = diagonal
```

## 2. Alice Generates Bits and Bases


```python
alice_bits = [random.randint(0, 1) for _ in range(NUM_BITS)]
alice_bases = [random.choice(BASES) for _ in range(NUM_BITS)]

list(zip(alice_bits, alice_bases))[:10]
```

## 3. Bob Chooses Random Bases

If Bob uses the same basis as Alice, he obtains the correct bit. If he uses a different basis, the result is random.


```python
bob_bases = [random.choice(BASES) for _ in range(NUM_BITS)]

bob_results = []
for bit, alice_basis, bob_basis in zip(alice_bits, alice_bases, bob_bases):
    if alice_basis == bob_basis:
        bob_results.append(bit)
    else:
        bob_results.append(random.randint(0, 1))

list(zip(alice_bits, alice_bases, bob_bases, bob_results))[:10]
```

## 4. Sifting

Alice and Bob publicly compare bases and keep only the positions where their bases match.


```python
matching_positions = [i for i, (a, b) in enumerate(zip(alice_bases, bob_bases)) if a == b]
alice_key = [alice_bits[i] for i in matching_positions]
bob_key = [bob_results[i] for i in matching_positions]

print('Matching positions:', matching_positions)
print('Alice key:', alice_key)
print('Bob key:  ', bob_key)
print('Keys match:', alice_key == bob_key)
```

## 5. Cybersecurity Relevance

QKD helps establish shared keys using quantum communication principles. It is different from Post-Quantum Cryptography.

- QKD requires specialized quantum communication infrastructure.
- PQC uses classical algorithms designed to resist quantum attacks.

For most software and network deployments, PQC is currently the more practical migration path.
