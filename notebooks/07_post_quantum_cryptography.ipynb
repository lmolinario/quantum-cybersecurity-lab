This notebook summarizes the cybersecurity relevance of Post-Quantum Cryptography (PQC).

The goal is not to implement production cryptography, but to understand the migration problem from a security architecture and governance perspective.


## 1. Why Post-Quantum Cryptography Matters

Post-Quantum Cryptography studies cryptographic algorithms designed to resist both classical and quantum adversaries.

The main concern is that sufficiently powerful fault-tolerant quantum computers could threaten widely used public-key cryptography.

## 3. Harvest Now, Decrypt Later

The most relevant near-term strategic concern is not that current quantum computers can already break modern cryptography at scale.

The concern is that adversaries may collect encrypted data today and decrypt it in the future if suitable quantum capabilities become available.

This matters especially for data with long confidentiality lifetimes.

## 5. QKD vs PQC

Quantum Key Distribution and Post-Quantum Cryptography are different approaches.

- QKD requires specialized quantum communication infrastructure.
- PQC uses classical software and protocols designed to resist quantum attacks.

For most organizations, PQC is the more practical migration path.


## 6. Key Takeaways

- PQC is a cybersecurity migration problem, not only a mathematical problem.
- The highest-priority systems are those with long-term confidentiality requirements.
- RSA, ECC, and classical key exchange mechanisms require strategic attention.
- Symmetric cryptography is affected differently and can often be mitigated with larger key sizes.
- PQC readiness requires inventory, architecture, governance, and testing.
