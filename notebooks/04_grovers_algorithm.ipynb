# 04 - Grover's Algorithm

This notebook demonstrates a small Grover-style search over a two-qubit search space.

The goal is educational: show how an oracle and diffusion step can amplify a target state.


````python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

simulator = AerSimulator()
````

## 1. Search Space

With two qubits, the search space contains four states:

```text
00
01
10
11
```

In this toy example, the target state is `11`.


## 2. Build the Toy Grover Circuit


````python
qc = QuantumCircuit(2, 2)

# Step 1: create uniform superposition
qc.h([0, 1])

# Step 2: oracle marking |11>
qc.cz(0, 1)

# Step 3: diffusion operator
qc.h([0, 1])
qc.x([0, 1])
qc.h(1)
qc.cx(0, 1)
qc.h(1)
qc.x([0, 1])
qc.h([0, 1])

qc.measure([0, 1], [0, 1])

print(qc)
````

## 3. Run the Circuit


````python
result = simulator.run(qc, shots=1024).result()
counts = result.get_counts()
counts
````

Expected result: the target state `11` should dominate the measurement counts.


## 4. Cybersecurity Relevance

Grover's algorithm provides a quadratic speedup for unstructured search.

For cybersecurity, this affects brute-force security margins for:

- symmetric keys
- password search
- hash preimage search

It does not break symmetric cryptography in the same way Shor threatens RSA or ECC, but it motivates larger key sizes for long-term security.
