## 1. Classical Bit vs Qubit

A classical bit can be either `0` or `1`.

A qubit can be represented as a quantum state. Before measurement, it can behave as a combination of the computational basis states `|0>` and `|1>`.

After measurement, the result becomes classical: either `0` or `1`.


```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

simulator = AerSimulator()

print('Qiskit and AerSimulator imported successfully.')
```

## 2. Measuring a Qubit Initialized in `|0>`

By default, a qubit starts in the `|0>` state. If we measure it immediately, we expect the result to be `0`.


```python
qc = QuantumCircuit(1, 1)
qc.measure(0, 0)

print(qc)

result = simulator.run(qc, shots=1024).result()
counts = result.get_counts()
counts
```

Expected result: most or all measurements should return `0`.


## 3. The X Gate

The X gate is similar to a classical NOT gate. It transforms `|0>` into `|1>` and `|1>` into `|0>`.


```python
qc = QuantumCircuit(1, 1)
qc.x(0)
qc.measure(0, 0)

print(qc)

result = simulator.run(qc, shots=1024).result()
counts = result.get_counts()
counts
```

Expected result: most or all measurements should return `1`.


## 4. Hadamard Gate and Superposition

The Hadamard gate creates a superposition. Starting from `|0>`, applying `H` creates a state that has approximately equal probability of being measured as `0` or `1`.


```python
qc = QuantumCircuit(1, 1)
qc.h(0)
qc.measure(0, 0)

print(qc)

result = simulator.run(qc, shots=1024).result()
counts = result.get_counts()
counts
```

Expected result: approximately 50% `0` and 50% `1`. The exact values change because measurement is probabilistic.


## 5. Bell State: First Example of Entanglement

A Bell state is one of the simplest examples of quantum entanglement. It can be created using a Hadamard gate followed by a CNOT gate.


```python
qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
qc.measure([0, 1], [0, 1])

print(qc)

result = simulator.run(qc, shots=1024).result()
counts = result.get_counts()
counts
```

Expected result: mostly `00` and `11`. This indicates correlated measurement outcomes.


## 6. Cybersecurity Relevance

Quantum computing matters for cybersecurity because some quantum algorithms affect assumptions used in classical cryptography.

Examples:

- Shor's algorithm threatens RSA and elliptic-curve cryptography if sufficiently powerful fault-tolerant quantum computers become available.
- Grover's algorithm reduces brute-force security margins through a quadratic speedup.
- Post-Quantum Cryptography studies classical algorithms designed to resist both classical and quantum attacks.

This repository uses simple Qiskit experiments to build intuition before moving to cybersecurity-specific analysis.
