```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

simulator = AerSimulator()
```

## 1. X Gate

The X gate flips a qubit. It is similar to a classical NOT operation.


```python
qc = QuantumCircuit(1, 1)
qc.x(0)
qc.measure(0, 0)

print(qc)
counts = simulator.run(qc, shots=1024).result().get_counts()
counts
```

## 2. Hadamard Gate

The Hadamard gate creates a superposition and is one of the most frequently used gates in quantum algorithms.


```python
qc = QuantumCircuit(1, 1)
qc.h(0)
qc.measure(0, 0)

print(qc)
counts = simulator.run(qc, shots=1024).result().get_counts()
counts
```

## 3. Z Gate and Phase

The Z gate changes the phase of the `|1>` component. Phase is not directly visible in a simple computational-basis measurement, but it affects later interference.


```python
qc = QuantumCircuit(1, 1)
qc.h(0)
qc.z(0)
qc.h(0)
qc.measure(0, 0)

print(qc)
counts = simulator.run(qc, shots=1024).result().get_counts()
counts
```

## 4. CNOT Gate

The CNOT gate uses one qubit as a control and another as a target. If the control is `1`, the target is flipped.


```python
qc = QuantumCircuit(2, 2)
qc.x(0)
qc.cx(0, 1)
qc.measure([0, 1], [0, 1])

print(qc)
counts = simulator.run(qc, shots=1024).result().get_counts()
counts
```

## 5. Security Relevance

Quantum gates are the building blocks of algorithms such as Grover and Shor. Understanding gates is necessary before studying the impact of quantum computing on cryptography.
