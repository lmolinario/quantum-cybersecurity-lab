```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

simulator = AerSimulator()
```

## 1. Bell Phi Plus

The Bell state `|Phi+>` is commonly written as `( |00> + |11> ) / sqrt(2)`.


```python
qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
qc.measure([0, 1], [0, 1])

print(qc)
counts = simulator.run(qc, shots=1024).result().get_counts()
counts
```

Expected measurement outcomes are mostly `00` and `11`.


## 2. Correlation

The two qubits are correlated: measuring one qubit gives information about the expected measurement result of the other qubit.

This does not allow faster-than-light communication, but it is a core resource for quantum information.


## 3. Bell States and Phase

Some Bell states differ by relative phase. If measured only in the computational basis, states that differ only by phase may produce the same measurement counts.

Distinguishing those states requires additional basis changes or statevector analysis.


## 4. Cybersecurity Relevance

Entanglement is relevant to quantum communication and quantum key distribution. It is also one of the principles that makes quantum computing different from classical computing.
